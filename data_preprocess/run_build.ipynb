```python
from utils.format import format_prompt
from datasets import Dataset

data_source = 'test'
output_style = format_prompt_kwargs.get('output_style', 'math_reward_style')
split = 'test'
ability = 'math'

qa_pairs = [
	{"question": "Calculate 1+2+...+100", "answer": "5050"
	},
] * 1

data_list = []
for idx, qa_pair in enumerate(qa_pairs):
	question = qa_pair['question']
	ground_truth = qa_pair['answer']
	prompt = format_prompt(question, **format_prompt_kwargs)
	data = {
		"data_source": f'{output_style}/{data_source}',
		"prompt": prompt,
		"ability": ability,
		"reward_model": {"style": "rule", "ground_truth": ground_truth},
		"extra_info": {"split": split, "index": idx, },
	}
	data_list.append(data)
dataset = Dataset.from_list(data_list)

dataset.to_parquet(os.path.join(local_save_dir, 'debug_testcase', f"test.parquet"))
```
